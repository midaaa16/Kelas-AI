2. Minimal Theory — Dasar Teori
Metode HVSR menggunakan perbandingan spektrum getaran horizontal dan vertikal. Jika komponen horizontal dinyatakan sebagai $(H_{NS}(f))$ dan $(H_{EW}(f))$, salah satu formulasi yang umum digunakan adalah:


$$H(f) =\sqrt{\frac{H_{NS}^{2}(f) + H_{EW}^{2}(f)}{2}}$$

$$HVSR(f)=\frac{H(f)}{|H_V(f)|}$$

Keterangan:
- $(H_{NS}(f))$: spektrum komponen horizontal utara–selatan.
- $(H_{EW}(f))$: spektrum komponen horizontal timur–barat.
- $(H_V(f))$: spektrum komponen vertikal.
- $(f)$: frekuensi dalam Hz.
Dari kurva HVSR, dua parameter utama yang dapat diperoleh adalah frekuensi dominan $(f_0)$ dan amplitudo puncak $(A_0)$.

Frekuensi dominan $(f_0)$ 

Frekuensi pada puncak utama kurva HVSR yang memenuhi kriteria pemilihan dan keandalan. Nilai ini dapat membantu mengidentifikasi variasi karakteristik tanah dan kontras impedansi bawah permukaan.




Amplitudo puncak $(A_0)$

Nilai HVSR pada frekuensi dominan. Nilai ini menggambarkan rasio spektral horizontal terhadap vertikal, tetapi tidak selalu sama dengan faktor amplifikasi gerakan tanah yang sebenarnya.




Periode dominan $(T_0)$
$$T_0=\frac{1}{f_0}$$

Periode dinyatakan dalam sekon. Periode yang lebih panjang dapat berkaitan dengan respons frekuensi rendah, tetapi interpretasinya tetap memerlukan konteks geologi.



Sebagai pemeriksaan tambahan, parameter indeks kerentanan seismik $(K_g)$  dapat dihitung menggunakan:

$$K_g=\frac{A_0^2}{f_0}$$

Penggunaan dan satuan $(K_g)$  harus mengikuti definisi yang digunakan dalam penelitian. Nilai ini bukan pengukuran langsung kerusakan gempa atau kerentanan bangunan.

Kriteria kualitas kurva, kestabilan puncak, dan pemilihan $(f_0)$ dapat mengacu pada pedoman SESAME. Kurva yang memiliki puncak tidak jelas atau tidak stabil perlu diberi peringatan dan tidak boleh langsung ditafsirkan sebagai frekuensi fundamental yang andal.

```python
for tr in st:
    print("Channel       :", tr.stats.channel)
    print("Sampling rate :", tr.stats.sampling_rate, "Hz")
    print("Jumlah sampel :", tr.stats.npts)
    print("Durasi        :", tr.stats.npts / tr.stats.sampling_rate, "detik")
    print("Waktu mulai   :", tr.stats.starttime)
    print("-" * 40)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.ndimage import gaussian_filter1d

# ==================================================
# A. MENYIAPKAN TIGA KOMPONEN
# ==================================================

tr_z = st.select(channel="BHZ")[0].copy()
tr_n = st.select(channel="BHN")[0].copy()
tr_e = st.select(channel="BHE")[0].copy()

# Memastikan sampling rate sama
assert tr_z.stats.sampling_rate == tr_n.stats.sampling_rate
assert tr_z.stats.sampling_rate == tr_e.stats.sampling_rate

# Memastikan waktu mulai dan akhir sama
assert tr_z.stats.starttime == tr_n.stats.starttime
assert tr_z.stats.starttime == tr_e.stats.starttime

assert tr_z.stats.npts == tr_n.stats.npts
assert tr_z.stats.npts == tr_e.stats.npts

fs = tr_z.stats.sampling_rate

z = tr_z.data.astype(float)
n = tr_n.data.astype(float)
e = tr_e.data.astype(float)

# Menghilangkan tren linear dan nilai rata-rata
z = signal.detrend(z, type="linear")
n = signal.detrend(n, type="linear")
e = signal.detrend(e, type="linear")

print("Preprocessing awal berhasil.")
print("Sampling rate:", fs, "Hz")
print("Durasi:", len(z) / fs, "detik")

# ==================================================
# B. PARAMETER ANALISIS
# ==================================================

window_seconds = 40
window_size = int(window_seconds * fs)
step_size = window_size // 2  # overlap 50%

psd_z = []
psd_n = []
psd_e = []

# ==================================================
# C. ANALISIS SETIAP JENDELA
# ==================================================

for start in range(0, len(z) - window_size + 1, step_size):

    end = start + window_size

    wz = z[start:end].copy()
    wn = n[start:end].copy()
    we = e[start:end].copy()

    # Menghilangkan tren dalam setiap jendela
    wz = signal.detrend(wz)
    wn = signal.detrend(wn)
    we = signal.detrend(we)

    # Taper Hann
    taper = np.hanning(window_size)

    wz = wz * taper
    wn = wn * taper
    we = we * taper

    # Estimasi spektrum daya
    f, pz = signal.periodogram(
        wz, fs=fs, window="boxcar",
        detrend=False, scaling="density"
    )
    _, pn = signal.periodogram(
        wn, fs=fs, window="boxcar",
        detrend=False, scaling="density"
    )
    _, pe = signal.periodogram(
        we, fs=fs, window="boxcar",
        detrend=False, scaling="density"
    )

    psd_z.append(pz)
    psd_n.append(pn)
    psd_e.append(pe)

# Merata-ratakan spektrum seluruh jendela
Pz = np.mean(psd_z, axis=0)
Pn = np.mean(psd_n, axis=0)
Pe = np.mean(psd_e, axis=0)

# Komponen horizontal gabungan
H = np.sqrt((Pn + Pe) / 2.0)

# HVSR berbasis spektrum daya
hvsr = np.full_like(H, np.nan)
valid = Pz > 0
hvsr[valid] = H[valid] / np.sqrt(Pz[valid])

print("Jumlah jendela:", len(psd_z))
print("Panjang jendela:", window_seconds, "detik")
print("Overlap: 50%")

# ==================================================
# D. PEMBATASAN RENTANG FREKUENSI
# ==================================================

f_min = 0.2
f_max = 20.0

mask = (
    (f >= f_min) &
    (f <= f_max) &
    np.isfinite(hvsr) &
    (hvsr > 0)
)

f_plot = f[mask]
h_plot = hvsr[mask]

if len(f_plot) == 0:
    print("Tidak ada data valid pada rentang frekuensi tersebut.")
else:
    # Penghalusan visual ringan; bukan validasi kualitas puncak
    h_smooth = gaussian_filter1d(h_plot, sigma=2)

    # Kandidat puncak tertinggi pada rentang yang dipilih
    idx = np.argmax(h_smooth)

    f0 = f_plot[idx]
    A0 = h_smooth[idx]
    T0 = 1.0 / f0

    plt.figure(figsize=(10, 6))
    plt.plot(f_plot, h_plot, alpha=0.45,
             label="HVSR awal")
    plt.plot(f_plot, h_smooth, linewidth=2,
             label="HVSR dihaluskan")

    plt.axvline(f0, linestyle="--",
                label="Kandidat f0 = {:.3f} Hz".format(f0))

    plt.scatter([f0], [A0], zorder=3)

    plt.xscale("log")
    plt.xlabel("Frekuensi (Hz)")
    plt.ylabel("HVSR")
    plt.title("Kurva HVSR Awal - Titik01")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    print("Kandidat frekuensi dominan f0:",
          round(f0, 4), "Hz")
    print("Kandidat amplitudo A0:",
          round(A0, 4))
    print("Periode dominan T0:",
          round(T0, 4), "detik")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter1d

# Rentang frekuensi analisis
f_min = 0.2
f_max = 20.0

# Menghitung HVSR masing-masing jendela
hvsr_windows = []

for pn_i, pe_i, pz_i in zip(psd_n, psd_e, psd_z):
    horizontal = np.sqrt((pn_i + pe_i) / 2.0)
    vertical = np.sqrt(pz_i)

    ratio = np.full_like(horizontal, np.nan, dtype=float)
    valid = vertical > 0
    ratio[valid] = horizontal[valid] / vertical[valid]

    hvsr_windows.append(ratio)

hvsr_windows = np.asarray(hvsr_windows)

mask = (
    (f >= f_min) &
    (f <= f_max)
)

freq = f[mask]
curves = hvsr_windows[:, mask]

# Statistik antarkurva pada setiap frekuensi
median_curve = np.nanmedian(curves, axis=0)
p16 = np.nanpercentile(curves, 16, axis=0)
p84 = np.nanpercentile(curves, 84, axis=0)

# Penghalusan untuk membantu identifikasi kandidat puncak
smooth_curve = gaussian_filter1d(median_curve, sigma=2)

# Kandidat puncak tertinggi
idx = np.nanargmax(smooth_curve)
f0_check = freq[idx]
A0_check = smooth_curve[idx]
T0_check = 1.0 / f0_check

# Grafik median dan sebaran antarkurva
plt.figure(figsize=(11, 6))

plt.fill_between(
    freq, p16, p84,
    alpha=0.2,
    label="Rentang persentil 16–84"
)

plt.plot(
    freq, median_curve,
    alpha=0.6,
    label="Median HVSR per jendela"
)

plt.plot(
    freq, smooth_curve,
    linewidth=2,
    label="Median HVSR dihaluskan"
)

plt.axvline(
    f0_check,
    linestyle="--",
    label="Kandidat f0 = {:.3f} Hz".format(f0_check)
)

plt.xscale("log")
plt.xlabel("Frekuensi (Hz)")
plt.ylabel("HVSR")
plt.title("Pemeriksaan Kestabilan Kurva HVSR - Titik01")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print("Jumlah jendela:", len(hvsr_windows))
print("Kandidat f0 hasil median:", round(f0_check, 4), "Hz")
print("Kandidat A0 hasil median:", round(A0_check, 4))
print("Kandidat T0 hasil median:", round(T0_check, 4), "detik")
```
